# Fetcher for the Strategies - Boiler plate

```python
from achest import MarketDataClient
client = MarketDataClient("https://achestv2.misango.me")
```

### For the Graham Buffett strat

```python
data = client.download(
["AAPL", "AXP", "KO", "BAC", "GOOGL", "CVX", "OXY", "MCO", "CB", "KHC"],
"2015-01-01",
"2026-11-07",
resolution="daily",
format="lean",
output="data/equity/usa/daily")
```
